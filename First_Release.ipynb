```python
import re

def validar_password(password):
    tiene_min_length = len(password) >= 8
    tiene_especial = bool(re.search(r'[!@#$%^&*(),.?":{}|<>]', password))

    return tiene_min_length and tiene_especial


# Pedir contraseña al usuario
password = input("Escribe tu contraseña: ")

if validar_password(password):
    print("✅ Contraseña válida y segura")
else:
    print("❌ Contraseña no cumple los requisitos")
```

```python
import getpass
import re # Importamos el módulo de expresiones regulares para la validación de contraseñas
import random # Importamos el módulo random para generar códigos de desbloqueo

# Diccionario para almacenar información de usuarios: contraseña, intentos fallidos, estado de bloqueo
usuarios_registrados = {
    "MateX775": {"password": "12pp34pp", "failed_attempts": 0, "locked": False} # Usuario inicial
}

def validar_contrasena(contrasena):
    if len(contrasena) < 8:
        print("❌ La contraseña debe tener al menos 8 caracteres.")
        return False
    if not re.search(r"[a-z]", contrasena):
        print("❌ La contraseña debe contener al menos una letra minúscula.")
        return False
    if not re.search(r"[0-9]", contrasena):
        print("❌ La contraseña debe contener al menos un número.")
        return False
    if not re.search(r"[!@#$%^&*(),.?\":{}|<>]", contrasena):
        print("❌ La contraseña debe contener al menos un carácter especial.")
        return False
    return True

def registrar_usuario():
    print("\n--- REGISTRO DE NUEVO USUARIO ---")
    while True:
        nuevo_usuario = input("Ingresa un nuevo nombre de usuario: ").strip()
        if not nuevo_usuario:
            print("❌ El nombre de usuario no puede estar vacío. Inténtalo de nuevo.")
        elif nuevo_usuario in usuarios_registrados:
            print("❌ Este nombre de usuario ya existe. Por favor, elige otro.")
        else:
            break

    while True:
        nueva_contrasena = getpass.getpass("Ingresa una nueva contraseña: ").strip()
        confirmar_contrasena = getpass.getpass("Confirma la contraseña: ").strip()
        if not nueva_contrasena:
            print("❌ La contraseña no puede estar vacía. Inténtalo de nuevo.")
        elif nueva_contrasena != confirmar_contrasena:
            print("❌ Las contraseñas no coinciden. Inténtalo de nuevo.")
        elif not validar_contrasena(nueva_contrasena):
            # La función validar_contrasena ya imprime el mensaje de error
            pass
        else:
            break

    # Almacenar la información completa del usuario
    usuarios_registrados[nuevo_usuario] = {"password": nueva_contrasena, "failed_attempts": 0, "locked": False}
    print(f"✅ Usuario '{nuevo_usuario}' registrado exitosamente.")

def iniciar_sesion():
    print("\n--- INICIO DE SESIÓN ---")
    usuario_ingresado = input("Ingresa tu nombre de usuario: ").strip()

    if usuario_ingresado not in usuarios_registrados:
        print("❌ Nombre de usuario o contraseña incorrectos.")
        print("🚨 ¡Alerta de seguridad! Se ha reportado un intento de inicio de sesión fallido.")
        return False

    user_data = usuarios_registrados[usuario_ingresado]

    if user_data["locked"]:
        if "temp_unlock_code" in user_data:
            print(f"❌ La cuenta de '{usuario_ingresado}' está bloqueada. Ingresa el código de 4 dígitos proporcionado por el administrador para desbloquearla.")
            codigo_ingresado = input("Ingresa el código de 4 dígitos: ").strip()
            if codigo_ingresado == user_data["temp_unlock_code"]:
                user_data["locked"] = False
                user_data["failed_attempts"] = 0
                del user_data["temp_unlock_code"]
                print(f"✅ Cuenta de '{usuario_ingresado}' desbloqueada exitosamente. Intentando iniciar sesión de nuevo...")
                # Ahora que está desbloqueada, el usuario debería ingresar su contraseña normal
                # O, para simplificar, si el código es correcto, iniciamos sesión directamente.
                print(f"✅ ¡Inicio de sesión exitoso! Bienvenido/a, {usuario_ingresado}.")
                return True
            else:
                print("❌ Código de desbloqueo incorrecto. La cuenta sigue bloqueada.")
                print("🚨 ¡Alerta de seguridad! Intento de desbloqueo fallido.")
                return False
        else:
            print(f"❌ La cuenta de '{usuario_ingresado}' está bloqueada debido a demasiados intentos fallidos. Contacta al administrador para obtener un código de desbloqueo.")
            return False

    contrasena_ingresada = getpass.getpass("Ingresa tu contraseña: ").strip()
    if user_data["password"] == contrasena_ingresada:
        print(f"✅ ¡Inicio de sesión exitoso! Bienvenido/a, {usuario_ingresado}.")
        user_data["failed_attempts"] = 0 # Reiniciar intentos fallidos al iniciar sesión correctamente
        return True
    else:
        user_data["failed_attempts"] += 1
        print("❌ Nombre de usuario o contraseña incorrectos.")
        print("🚨 ¡Alerta de seguridad! Se ha reportado un intento de inicio de sesión fallido.")

        if user_data["failed_attempts"] >= 3:
            user_data["locked"] = True
            print(f"⛔ La cuenta de '{usuario_ingresado}' ha sido bloqueada debido a 3 intentos fallidos.")
        else:
            print(f"Te quedan {3 - user_data['failed_attempts']} intentos antes de que tu cuenta sea bloqueada.")
        return False

def contactar_admin():
    print("\n--- CONTACTAR ADMINISTRADOR ---")
    print("Para conectar con el administrador, por favor ingresa el número de contacto secreto.")
    numero_secreto = "4438416054"
    admin_key = "1992"

    while True:
        intentos_restantes_numero = 3 # Intentos para el número secreto
        numero_ingresado = input("Ingresa el número: ").strip()
        if numero_ingresado == numero_secreto:
            print("✅ Conexión establecida. El administrador ha sido notificado.")
            print("📞 Espera su llamada...")

            # Nueva lógica para la clave del administrador
            print("\n--- ACCESO DE ADMINISTRADOR ---")
            print("Para acceder a las funciones de administrador, ingresa la clave.")

            intentos_restantes_clave = 3 # Intentos para la clave de admin
            while intentos_restantes_clave > 0:
                clave_ingresada = getpass.getpass("Ingresa la clave de administrador: ").strip()
                if clave_ingresada == admin_key:
                    print("✅ Clave de administrador correcta. Acceso concedido.")

                    # Mostrar cuentas bloqueadas y generar códigos de desbloqueo
                    print("\n--- CUENTAS BLOQUEADAS ---")
                    cuentas_bloqueadas = [user for user, data in usuarios_registrados.items() if data["locked"]]

                    if cuentas_bloqueadas:
                        print("Las siguientes cuentas están bloqueadas:")
                        for cuenta in cuentas_bloqueadas:
                            # Generar un código de 4 dígitos para cada cuenta bloqueada
                            unlock_code = str(random.randint(1000, 9999))
                            usuarios_registrados[cuenta]["temp_unlock_code"] = unlock_code
                            print(f" - {cuenta} (Código de desbloqueo temporal: {unlock_code})")
                    else:
                        print("No hay cuentas bloqueadas actualmente.")
                    return # Sale de la función contactar_admin
                else:
                    intentos_restantes_clave -= 1
                    print("❌ Clave de administrador incorrecta.")
                    if intentos_restantes_clave > 0:
                        print(f"Te quedan {intentos_restantes_clave} intentos para la clave de administrador.")
                    else:
                        print("⛔ Demasiados intentos fallidos para la clave de administrador. Acceso denegado.")
                        return # Sale de la función contactar_admin si falla la clave
            break # Sale del bucle principal de contactar_admin si se accede y luego se cierra

        else:
            intentos_restantes_numero -= 1
            print("1❌ Número incorrecto.")
            if intentos_restantes_numero <= 0:
                print("💀 Oh, te atreviste a molestarme... No debiste hacer eso.")
                print("👁️ Sé dónde estás y sé lo que hiciste.")
                print("🔪 El administrador no está disponible. Ahora tú eres el problema.")
                print("\u001B[31m¡Preparate para las consecuencias!\u001B[0m") # Red color for extra creepy effect
                break
            else:
                print(f"Aún te quedan {intentos_restantes_numero} intentos. Piensa bien.")

def mostrar_menu_principal():
    print("\n--- MENÚ PRINCIPAL ---")
    print("1. Iniciar Sesión")
    print("2. Registrar Usuario")
    print("3. Contactar Administrador")
    print("4. Salir")

def mostrar_menu_post_login(usuario):
    print(f"\n--- MENÚ DE SESIÓN para {usuario} ---")
    print("1. Ver perfil (simulado)")
    print("2. Cambiar contraseña (simulado)")
    print("3. Cerrar sesión")

    while True:
        opcion_post_login = input("Elige una opción: ").strip()
        if opcion_post_login == '1':
            print(f"\nInformación del perfil de {usuario}:")
            print(" - Rol: Usuario Estándar")
            print(" - Último inicio de sesión: Ahora mismo")
        elif opcion_post_login == '2':
            print("\nFunción para cambiar contraseña aún no implementada completamente.")
        elif opcion_post_login == '3':
            print("Cerrando sesión...")
            return # Sale del menú post-login
        else:
            print("❌ Opción no válida. Por favor, elige 1, 2 o 3.")

# Bucle principal del menú
while True:
    mostrar_menu_principal()
    opcion = input("Elige una opción: ").strip()

    if opcion == '1':
        # Modificar iniciar_sesion para que devuelva el nombre de usuario si es exitoso
        usuario_loggeado = iniciar_sesion()
        if usuario_loggeado:
            mostrar_menu_post_login(usuario_loggeado)
    elif opcion == '2':
        registrar_usuario()
    elif opcion == '3':
        contactar_admin()
    elif opcion == '4':
        print("¡Hasta luego!")
        break # Sale del bucle principal
    else:
        print("❌ Opción no válida. Por favor, elige 1, 2, 3 o 4.")
```

```python
# 1. Crear un diccionario
frutas_colores = {
    "manzana": "rojo",
    "banana": "amarillo",
    "uva": "morado",
    "limon": "verde"
}
print("Diccionario de frutas y colores:", frutas_colores)

# 2. Acceder a un valor usando su clave
print("\nEl color de la manzana es:", frutas_colores["manzana"])

# 3. Modificar un valor
frutas_colores["manzana"] = "verde y rojo"
print("\nDiccionario después de modificar la manzana:", frutas_colores)

# 4. Añadir un nuevo par clave-valor
frutas_colores["naranja"] = "naranja"
print("\nDiccionario después de añadir naranja:", frutas_colores)

# 5. Eliminar un par clave-valor
del frutas_colores["limon"]
print("\nDiccionario después de eliminar limon:", frutas_colores)

# 6. Iterar sobre las claves del diccionario
print("\nClaves del diccionario:")
for fruta in frutas_colores:
    print(fruta)

# 7. Iterar sobre los valores del diccionario
print("\nValores del diccionario:")
for color in frutas_colores.values():
    print(color)

# 8. Iterar sobre claves y valores (elementos)
print("\nPares clave-valor del diccionario:")
for fruta, color in frutas_colores.items():
    print(f"La {fruta} es {color}.")

# 9. Verificar si una clave existe en el diccionario
print("\n¿Está 'banana' en el diccionario?", "banana" in frutas_colores)
print("¿Está 'kiwi' en el diccionario?", "kiwi" in frutas_colores)
```

```python
# 1. Crear un diccionario a partir de una lista de tuplas (pares clave-valor)
datos_estudiantes = [
    ("MateX", 25),
    ("Ana", 30),
    ("Luis", 22)
]
estudiantes_edades = dict(datos_estudiantes)
print("\nDiccionario de estudiantes y edades:", estudiantes_edades)

# 2. Usar .get() para acceder a valores de forma segura (sin KeyError si la clave no existe)
# Si la clave existe, devuelve el valor
print("\nEdad de MateX (con get()):", estudiantes_edades.get("MateX"))
# Si la clave no existe, devuelve None por defecto o un valor especificado
print("Edad de Carlos (con get() y valor por defecto):", estudiantes_edades.get("Carlos", "No registrado"))

# 3. Actualizar un diccionario con otro diccionario o con pares clave-valor
otros_estudiantes = {"Pedro": 28, "Ana": 31} # Ana se actualiza
estudiantes_edades.update(otros_estudiantes)
print("\nDiccionario después de actualizar:", estudiantes_edades)

# 4. Obtener todas las claves como una lista
claves = list(estudiantes_edades.keys())
print("\nTodas las claves:", claves)

# 5. Obtener todos los valores como una lista
valores = list(estudiantes_edades.values())
print("\nTodos los valores:", valores)

# 6. Copiar un diccionario (copia superficial)
diccionario_copia = estudiantes_edades.copy()
diccionario_copia["Luis"] = 23 # Modificamos la copia
print("\nDiccionario original después de modificar la copia:", estudiantes_edades)
print("Diccionario copia:", diccionario_copia)
```

```python
import json
import os # Para verificar si un archivo existe

class Perro:
    ENIGMATIC_NAMES_LORE = {
        "SlenderDog": "Su mirada vacía te sigue incluso en la oscuridad. Sus ladridos no son de un perro, sino susurros de un secreto olvidado, un eco de la noche sin fin.",
        "Cerbero": "No es solo un perro, es el guardián de los umbrales. Sus tres cabezas no ladran, sino que entonan un cántico ancestral que congela el alma de quienes osan acercarse al velo.",
        "Krypto": "Un ser de otro mundo, su lealtad es inquebrantable, pero en sus ojos se esconde el pesar de un sol lejano, una estrella moribunda que lo dota de poderes más allá de la comprensión mortal."
        # Puedes añadir más nombres enigmáticos y su lore aquí
    }

    def __init__(self, nombre, raza, genero=None, tipo_elemental=None):
        self.nombre = nombre
        self.raza = raza
        self.genero = genero
        self.tipo_elemental = tipo_elemental

    def ladrar(self):
        if self.nombre in self.ENIGMATIC_NAMES_LORE:
            return f"{self.nombre} gime con un sonido que no es de este mundo..."
        return f"{self.nombre} dice: ¡Guau, guau!"

    def tipo_electrico(self):
        if self.tipo_elemental and self.tipo_elemental.lower() == "eléctrico":
            return f"{self.nombre} es un perro de tipo eléctrico y puede lanzar un pequeño chispazo."
        else:
            return f"{self.nombre} no es de tipo eléctrico."

    def presentarse(self):
        presentacion = f"Hola, soy {self.nombre}, un {self.raza}"
        if self.genero:
            if self.genero.lower() == "hembra":
                presentacion += " (hembra)"
            elif self.genero.lower() == "macho":
                presentacion += " (macho)"
        if self.tipo_elemental:
            presentacion += f" de tipo {self.tipo_elemental}."
        else:
            presentacion += "."

        if self.nombre in self.ENIGMATIC_NAMES_LORE:
            presentacion += f" {self.ENIGMATIC_NAMES_LORE[self.nombre]}"
        return presentacion

# --- Funciones de Guardado y Carga --- #
def save_perro_to_file(perro_obj):
    # Usamos el nombre del perro como parte del nombre del archivo
    filename = f"{perro_obj.nombre}.json"
    data = {
        'nombre': perro_obj.nombre,
        'raza': perro_obj.raza,
        'genero': perro_obj.genero,
        'tipo_elemental': perro_obj.tipo_elemental
    }
    try:
        with open(filename, 'w', encoding='utf-8') as f:
            json.dump(data, f, indent=4, ensure_ascii=False)
        print(f"✅ ¡Perro '{perro_obj.nombre}' guardado exitosamente en '{filename}'!")
    except Exception as e:
        print(f"❌ Error al guardar el perro: {e}")

def load_perro_from_file(nombre_perro_a_cargar):
    filename = f"{nombre_perro_a_cargar}.json"
    if not os.path.exists(filename):
        print(f"❌ No se encontró ningún archivo para '{nombre_perro_a_cargar}'.")
        return None
    try:
        with open(filename, 'r', encoding='utf-8') as f:
            data = json.load(f)
        print(f"✅ ¡Perro '{nombre_perro_a_cargar}' cargado exitosamente!")
        # Recrear el objeto Perro a partir de los datos cargados
        return Perro(data['nombre'], data['raza'], data['genero'], data['tipo_elemental'])
    except json.JSONDecodeError:
        print(f"❌ Error al leer el archivo '{filename}'. Asegúrate de que sea un JSON válido.")
        return None
    except Exception as e:
        print(f"❌ Error al cargar el perro: {e}")
        return None


# --- Parte interactiva --- #
print("\n--- Crea o Carga tu Perro --- ")
mi_perro = None

# Bucle para asegurar que se crea o carga un perro antes de continuar
while mi_perro is None:
    accion_inicial = input("¿Quieres (C)rear un nuevo perro o (L)oad uno existente? (C/L): ").lower()
    if accion_inicial == 'c':
        nombre_perro = input("Ingresa el nombre de tu perro: ")
        raza_perro = input("Ingresa la raza de tu perro: ")

        # Genero con una sola letra
        genero_input = input("Ingresa el género de tu perro (m=macho/h=hembra) o deja en blanco: ").lower()
        if genero_input == 'm':
            genero_perro = "macho"
        elif genero_input == 'h':
            genero_perro = "hembra"
        else:
            genero_perro = None

        # Tipo elemental con una sola letra
        tipo_input = input("¿Tu perro tiene algún tipo elemental? (ej: e=eléctrico, a=agua, f=fuego) o deja en blanco: ").lower()
        if tipo_input == 'e':
            tipo_perro = "eléctrico"
        elif tipo_input == 'a':
            tipo_perro = "agua"
        elif tipo_input == 'f':
            tipo_perro = "fuego"
        elif tipo_input: # Si no es una de las letras predefinidas, pero hay entrada, se guarda tal cual
            tipo_perro = tipo_input
        else:
            tipo_perro = None

        mi_perro = Perro(nombre_perro, raza_perro, genero_perro, tipo_perro)
        print(f"\n¡Felicidades! Has creado a {mi_perro.nombre}.")
    elif accion_inicial == 'l':
        nombre_a_cargar = input("Ingresa el nombre del perro que quieres cargar: ")
        mi_perro = load_perro_from_file(nombre_a_cargar)
        if mi_perro: # Si la carga fue exitosa, mi_perro no será None
            print(f"¡Bienvenido de nuevo, {mi_perro.nombre}!")
    else:
        print("Opción no válida. Por favor, elige 'C' o 'L'.")


# Menú de interacción con el perro
while True:
    print(f"\n--- ¿Qué quieres hacer con {mi_perro.nombre}? ---")
    print("1. Hacer que se presente")
    print("2. Hacer que ladre")
    print("3. Preguntar por su tipo eléctrico")
    print("4. Guardar los datos de tu perro")
    print("5. Salir")

    opcion = input("Elige una opción: ")

    if opcion == '1':
        print(mi_perro.presentarse())
    elif opcion == '2':
        print(mi_perro.ladrar())
    elif opcion == '3':
        print(mi_perro.tipo_electrico())
    elif opcion == '4':
        save_perro_to_file(mi_perro)
    elif opcion == '5':
        print(f"¡Adiós, {mi_perro.nombre}! No olvides guardar para la próxima.")
        break
    else:
        print("Opción no válida. Por favor, elige un número entre 1 y 5.")
```
